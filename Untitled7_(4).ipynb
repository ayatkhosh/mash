```python
# ================================================================
# LOAD CLIP FOR SEGMENT CLASSIFICATION
# ================================================================
# CLIP understands aerial imagery semantically:
#   "building rooftop from above" vs "green vegetation trees"
# Much better than HSV color rules!
# ================================================================

from transformers import CLIPProcessor, CLIPModel

CLIP_MODEL_ID = "openai/clip-vit-base-patch32"

print(f"Loading CLIP: {CLIP_MODEL_ID}...")
clip_model = CLIPModel.from_pretrained(CLIP_MODEL_ID)
clip_model = clip_model.to(device)
clip_model.eval()
clip_processor = CLIPProcessor.from_pretrained(CLIP_MODEL_ID)

print(f"CLIP loaded on {device.upper()}")
print(f"GPU Memory: {torch.cuda.memory_allocated()/1e9:.1f} GB")
```

```python
# ================================================================
# CLIP-BASED AERIAL SEGMENT CLASSIFIER
# ================================================================
# For each SAM segment:
#   1. Crop the image region around the mask
#   2. Apply mask to isolate the segment
#   3. Send to CLIP with descriptive text labels
#   4. CLIP returns the best matching category
#
# This replaces ALL HSV/color rules with semantic understanding.
# ================================================================

import torch
import numpy as np
import cv2
from PIL import Image as PILImage
from scipy import ndimage

# Text descriptions for each category — optimized for aerial/satellite view
CLIP_LABELS = [
    "aerial view of a building rooftop",
    "aerial view of green vegetation trees and grass",
    "aerial view of a road street or pavement",
    "aerial view of water river pool or lake",
    "aerial view of open bare land or empty lot",
]

CLIP_CATEGORIES = ['buildings', 'vegetation', 'road', 'water', 'open_land']


def classify_segment_clip(image_np, mask, min_crop_size=32):
    """
    Classify a segment using CLIP.
    Crops the region, applies mask, sends to CLIP.
    Returns: (category_name, confidence)
    """
    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)
    if not rows.any():
        return 'unknown', 0.0

    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]

    # Add padding around crop
    pad = 10
    h, w = image_np.shape[:2]
    rmin = max(0, rmin - pad)
    rmax = min(h - 1, rmax + pad)
    cmin = max(0, cmin - pad)
    cmax = min(w - 1, cmax + pad)

    crop = image_np[rmin:rmax+1, cmin:cmax+1].copy()
    mask_crop = mask[rmin:rmax+1, cmin:cmax+1]

    # Apply mask: darken non-segment pixels to help CLIP focus
    for ch in range(3):
        crop[:, :, ch] = np.where(mask_crop, crop[:, :, ch],
                                   (crop[:, :, ch] * 0.3).astype(np.uint8))

    # Skip very small crops
    if crop.shape[0] < min_crop_size or crop.shape[1] < min_crop_size:
        return 'unknown', 0.0

    crop_pil = PILImage.fromarray(crop)

    # CLIP inference
    inputs = clip_processor(
        text=CLIP_LABELS,
        images=crop_pil,
        return_tensors="pt",
        padding=True,
    ).to(device)

    with torch.no_grad():
        outputs = clip_model(**inputs)
        probs = outputs.logits_per_image.softmax(dim=1)[0]

    best_idx = probs.argmax().item()
    confidence = probs[best_idx].item()

    return CLIP_CATEGORIES[best_idx], confidence


def classify_segments_batch(image_np, sam_masks, batch_size=16):
    """
    Classify all SAM segments using CLIP in batches for speed.
    Returns dict with classified segments per category.
    """
    out = {c: [] for c in ['buildings', 'vegetation', 'road',
                            'open_land', 'water', 'unknown']}

    crops = []
    indices = []

    for idx, m in enumerate(sam_masks):
        mask = m['segmentation']
        area = m['area']

        if area < 30:
            continue

        rows = np.any(mask, axis=1)
        cols = np.any(mask, axis=0)
        if not rows.any():
            continue

        rmin, rmax = np.where(rows)[0][[0, -1]]
        cmin, cmax = np.where(cols)[0][[0, -1]]

        pad = 10
        h, w = image_np.shape[:2]
        rmin_p = max(0, rmin - pad)
        rmax_p = min(h - 1, rmax + pad)
        cmin_p = max(0, cmin - pad)
        cmax_p = min(w - 1, cmax + pad)

        crop = image_np[rmin_p:rmax_p+1, cmin_p:cmax_p+1].copy()
        mask_crop = mask[rmin_p:rmax_p+1, cmin_p:cmax_p+1]

        if crop.shape[0] < 20 or crop.shape[1] < 20:
            continue

        # Apply mask
        for ch in range(3):
            crop[:, :, ch] = np.where(mask_crop, crop[:, :, ch],
                                       (crop[:, :, ch] * 0.3).astype(np.uint8))

        crops.append(PILImage.fromarray(crop))
        indices.append(idx)

    print(f"  Classifying {len(crops)} segments with CLIP...")

    # Process in batches
    for batch_start in range(0, len(crops), batch_size):
        batch_end = min(batch_start + batch_size, len(crops))
        batch_crops = crops[batch_start:batch_end]
        batch_indices = indices[batch_start:batch_end]

        inputs = clip_processor(
            text=CLIP_LABELS,
            images=batch_crops,
            return_tensors="pt",
            padding=True,
        ).to(device)

        with torch.no_grad():
            outputs = clip_model(**inputs)
            logits = outputs.logits_per_image  # (batch, num_labels)
            probs = logits.softmax(dim=1)

        for i, idx in enumerate(batch_indices):
            m = sam_masks[idx]
            mask = m['segmentation']
            area = m['area']
            iou = m['predicted_iou']

            best_label_idx = probs[i].argmax().item()
            confidence = probs[i][best_label_idx].item()
            cat = CLIP_CATEGORIES[best_label_idx]

            shape = get_shape_info(mask)

            # Post-validation for buildings
            if cat == 'buildings':
                h_img, w_img = image_np.shape[:2]
                total_px = h_img * w_img
                if area / total_px > 0.25:
                    cat = 'open_land'
                elif shape['aspect_ratio'] > 15:
                    cat = 'road'
                elif shape['area'] < 80:
                    cat = 'unknown'

            out[cat].append({
                'mask': mask,
                'area': area,
                'quality': iou,
                'clip_score': confidence,
                'bbox': shape['bbox'],
                'centroid': shape['centroid'],
                'shape': shape,
            })

        if (batch_end) % 64 == 0 or batch_end == len(crops):
            print(f"    Processed {batch_end}/{len(crops)}")

    for cat, segs in out.items():
        if segs:
            print(f"  {cat:<14}: {len(segs):>4}")

    return out


def get_shape_info(mask):
    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)
    if not rows.any():
        return {'aspect_ratio': 1, 'rectangularity': 0, 'area': 0,
                'bbox': [0, 0, 0, 0], 'centroid': (0, 0), 'compactness': 0}
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    bh = max(rmax - rmin + 1, 1)
    bw = max(cmax - cmin + 1, 1)
    area = int(mask.sum())
    cy, cx = ndimage.center_of_mass(mask)
    contours, _ = cv2.findContours(
        mask.astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    perimeter = sum(cv2.arcLength(c, True) for c in contours) if contours else 1
    compactness = (4 * np.pi * area) / (perimeter**2 + 1e-6)
    return {
        'aspect_ratio': max(bh, bw) / min(bh, bw),
        'rectangularity': area / (bh * bw),
        'area': area,
        'bbox': [int(cmin), int(rmin), int(cmax), int(rmax)],
        'centroid': (int(cx), int(cy)),
        'compactness': float(min(compactness, 1.0)),
    }


print("CLIP classifier ready!")
print("Labels:", CLIP_CATEGORIES)
```

```python
# ================================================================
# SAM AUTO-SEGMENTATION + LAND USE MAP
# ================================================================

from segment_anything import SamAutomaticMaskGenerator


def run_sam_auto(image_np, points_per_side=48,
                 pred_iou_thresh=0.76, stability_thresh=0.80,
                 min_area=100):
    gen = SamAutomaticMaskGenerator(
        model=sam_model,
        points_per_side=points_per_side,
        pred_iou_thresh=pred_iou_thresh,
        stability_score_thresh=stability_thresh,
        min_mask_region_area=min_area,
        crop_n_layers=1,
        crop_n_points_downscale_factor=2,
    )
    with torch.no_grad():
        masks = gen.generate(image_np)
    return sorted(masks, key=lambda x: x['area'], reverse=True)


def build_land_use_map(image_np, classified):
    h, w = image_np.shape[:2]
    lmap = np.zeros((h, w), dtype=np.uint8)

    # Layer by priority (lowest first, highest last)
    for seg in classified.get('open_land', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 0
    for seg in classified.get('road', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 0
    for seg in classified.get('vegetation', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 1
    for seg in classified.get('water', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 3
    # Buildings = highest priority
    for seg in classified.get('buildings', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 2

    # Morphological cleanup
    k5 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
    bld_mask = (lmap == 2)
    veg_clean = cv2.morphologyEx((lmap == 1).astype(np.uint8), cv2.MORPH_OPEN, k5)
    veg_clean = cv2.morphologyEx(veg_clean, cv2.MORPH_CLOSE, k5)
    wat_clean = cv2.morphologyEx((lmap == 3).astype(np.uint8), cv2.MORPH_OPEN, k5)

    lmap[:] = 0
    lmap[veg_clean > 0] = 1
    lmap[wat_clean > 0] = 3
    lmap[bld_mask] = 2

    return lmap


print("SAM Auto + Land Use ready!")
```

```python
import cv2, numpy as np, uuid
from dataclasses import dataclass
from typing import List, Tuple
from PIL import Image as PILImage


@dataclass
class Building:
    building_id    : str
    bbox           : List[int]
    area_pixels    : int
    confidence     : float
    mask           : np.ndarray
    centroid       : Tuple[int, int]
    compactness    : float = 0.0
    rectangularity : float = 0.0


def clean_masks(masks, opening=3, closing=7, min_area=120):
    out = []
    for m in masks:
        area = m.sum()
        u = m.astype(np.uint8) * 255
        op_sz = max(2, min(opening, int(np.sqrt(area) / 10)))
        cl_sz = max(3, min(closing, int(np.sqrt(area) / 5)))
        k1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (op_sz, op_sz))
        k2 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (cl_sz, cl_sz))
        u = cv2.morphologyEx(u, cv2.MORPH_OPEN, k1)
        u = cv2.morphologyEx(u, cv2.MORPH_CLOSE, k2)
        result = u > 0
        if result.sum() >= min_area:
            out.append(result)
    return out


def merge_overlapping(masks, iou_threshold=0.3):
    if len(masks) <= 1:
        return masks
    merged = []
    used = set()
    for i in range(len(masks)):
        if i in used:
            continue
        current = masks[i].copy()
        for j in range(i + 1, len(masks)):
            if j in used:
                continue
            inter = np.logical_and(current, masks[j]).sum()
            union = np.logical_or(current, masks[j]).sum()
            if inter / max(union, 1) > iou_threshold:
                current = np.logical_or(current, masks[j])
                used.add(j)
        merged.append(current)
        used.add(i)
    return merged


def register_buildings(classified_blds):
    blds = []
    for seg in classified_blds:
        mask = seg['mask']
        shape = seg.get('shape', get_shape_info(mask))
        if shape['area'] < 50:
            continue
        bid = f'BLD-{str(uuid.uuid4())[:6].upper()}'
        PILImage.fromarray(mask.astype(np.uint8) * 255).save(
            DIRS['masks'] / f'{bid}.png')
        blds.append(Building(
            building_id=bid, bbox=shape['bbox'],
            area_pixels=shape['area'],
            confidence=float(seg.get('clip_score', seg.get('quality', 0))),
            mask=mask, centroid=shape['centroid'],
            compactness=shape.get('compactness', 0),
            rectangularity=shape.get('rectangularity', 0),
        ))
    blds.sort(key=lambda b: b.area_pixels, reverse=True)
    return blds


print("Post-processing ready!")
```

```python
from dataclasses import dataclass
from typing import List, Dict

@dataclass
class Neighborhood:
    grid_id: str
    row: int
    col: int
    bbox: List[int]
    n_buildings: int
    building_ids: List[str]
    total_pixels: int
    land_use_pct: Dict[str, float]
    density_class: str

def analyze_neighborhoods(land_map, buildings, grid_rows=3, grid_cols=3):
    h, w = land_map.shape
    cell_h, cell_w = h // grid_rows, w // grid_cols
    nbs = []
    for r in range(grid_rows):
        for c in range(grid_cols):
            y1 = r * cell_h
            y2 = (r + 1) * cell_h if r < grid_rows - 1 else h
            x1 = c * cell_w
            x2 = (c + 1) * cell_w if c < grid_cols - 1 else w
            cell = land_map[y1:y2, x1:x2]
            tp = cell.size
            pct = {
                'open': float(np.sum(cell == 0) / tp * 100),
                'vegetation': float(np.sum(cell == 1) / tp * 100),
                'built': float(np.sum(cell == 2) / tp * 100),
                'water': float(np.sum(cell == 3) / tp * 100),
            }
            bids = [b.building_id for b in buildings
                    if x1 <= b.centroid[0] < x2 and y1 <= b.centroid[1] < y2]
            density = 'high' if pct['built'] > 50 else 'medium' if pct['built'] > 20 else 'low'
            nbs.append(Neighborhood(
                grid_id=f"R{r}C{c}", row=r, col=c, bbox=[x1, y1, x2, y2],
                n_buildings=len(bids), building_ids=bids,
                total_pixels=tp, land_use_pct=pct, density_class=density))
    return nbs

def print_neighborhood_report(nbs):
    print("=" * 70)
    print("NEIGHBORHOOD ANALYSIS REPORT")
    print("=" * 70)
    for nb in nbs:
        p = nb.land_use_pct
        print(f"\n[{nb.grid_id}] {nb.density_class.upper()} | Bld: {nb.n_buildings} | "
              f"Built: {p['built']:.1f}% | Veg: {p['vegetation']:.1f}% | Open: {p['open']:.1f}%")
    print()

print("Neighborhood analysis ready!")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
plt.rcParams['font.family'] = 'DejaVu Sans'

def colored_overlay(image_np, masks, alpha=0.62):
    ov = image_np.copy().astype(float)
    cmap = plt.get_cmap('tab20')
    for i, m in enumerate(masks):
        if m.shape != image_np.shape[:2]: continue
        c = np.array(cmap(i % 20)[:3]) * 255
        for ch in range(3):
            ov[:, :, ch][m] = ov[:, :, ch][m] * (1 - alpha) + c[ch] * alpha
    return ov.astype(np.uint8)

def draw_hline(ax, y, xmin=0.05, xmax=0.95, color='#444', lw=1):
    ax.plot([xmin, xmax], [y, y], color=color, linewidth=lw,
            transform=ax.transAxes, clip_on=False, solid_capstyle='butt')

def visualize_results(image_np, classified, buildings, land_map,
                      neighborhoods=None, title='UrbanInsight SAM+CLIP', save_path=None):
    fig = plt.figure(figsize=(26, 17))
    fig.patch.set_facecolor('#0d1117')
    n_b = len(buildings)
    n_v = len(classified.get('vegetation', []))
    n_w = len(classified.get('water', []))
    n_r = len(classified.get('road', []))
    total_seg = sum(len(classified.get(c, [])) for c in
                    ['buildings', 'vegetation', 'road', 'water', 'open_land', 'unknown'])
    fig.suptitle(f'{title}  |  Buildings: {n_b}  Veg: {n_v}  Road: {n_r}  Water: {n_w}  Total: {total_seg}',
                 fontsize=15, fontweight='bold', color='white', y=0.99)
    gs = fig.add_gridspec(2, 3, hspace=0.28, wspace=0.14,
                          left=0.02, right=0.98, top=0.94, bottom=0.02)
    axes = [fig.add_subplot(gs[r, c]) for r in range(2) for c in range(3)]
    h, w = image_np.shape[:2]
    axes[0].imshow(image_np)
    axes[0].set_title('Original Image', color='white', fontsize=13, pad=8)

    CAT_COLORS = {'buildings': (220,70,50), 'vegetation': (50,195,70),
                  'road': (150,150,150), 'water': (60,140,220),
                  'open_land': (180,155,90), 'unknown': (100,100,100)}
    ov1 = image_np.copy().astype(float)
    leg1 = []
    for cat, (cr, cg, cb) in CAT_COLORS.items():
        segs = classified.get(cat, [])
        if not segs: continue
        c = np.array([cr, cg, cb], dtype=float)
        for seg in segs:
            m = seg['mask']
            if m.shape != (h, w): continue
            for ch in range(3):
                ov1[:, :, ch][m] = ov1[:, :, ch][m] * 0.25 + c[ch] * 0.75
        leg1.append(mpatches.Patch(color=np.array([cr,cg,cb])/255,
                                   label=f'{cat} ({len(segs)})'))
    axes[1].imshow(ov1.astype(np.uint8))
    if leg1:
        axes[1].legend(handles=leg1, loc='upper right', fontsize=8,
                       framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[1].set_title(f'SAM+CLIP Classification -- {total_seg} segments',
                      color='white', fontsize=13, pad=8)

    bld_masks = [b.mask for b in buildings]
    axes[2].imshow(colored_overlay(image_np, bld_masks, 0.70))
    axes[2].set_title(f'Building Masks -- {n_b}', color='white', fontsize=13, pad=8)

    lrgb = np.zeros((h, w, 3), dtype=np.uint8)
    lrgb[land_map == 0] = [90, 110, 200]
    lrgb[land_map == 1] = [50, 190, 70]
    lrgb[land_map == 2] = [210, 55, 55]
    lrgb[land_map == 3] = [55, 145, 220]
    total_px = land_map.size
    pct = {k: np.sum(land_map == v) / total_px * 100
           for k, v in {'open': 0, 'veg': 1, 'built': 2, 'water': 3}.items()}
    axes[3].imshow(lrgb)
    lu_leg = [
        mpatches.Patch(color='#5A6EC8', label=f"Open/Road {pct['open']:.1f}%"),
        mpatches.Patch(color='#32BE46', label=f"Vegetation {pct['veg']:.1f}%"),
        mpatches.Patch(color='#D23737', label=f"Built-up {pct['built']:.1f}%"),
        mpatches.Patch(color='#3791DC', label=f"Water {pct['water']:.1f}%"),
    ]
    axes[3].legend(handles=lu_leg, loc='upper right', fontsize=9,
                   framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[3].set_title('Land Use Map (SAM+CLIP)', color='white', fontsize=13, pad=8)

    axes[4].imshow(colored_overlay(image_np, bld_masks, 0.55))
    for b in buildings:
        cx, cy = b.centroid
        fs = max(5, min(9, int(b.area_pixels**0.38 * 0.30)))
        axes[4].text(cx, cy, b.building_id[-6:], color='white', fontsize=fs,
                     ha='center', va='center', fontweight='bold',
                     bbox=dict(boxstyle='round,pad=0.15', fc='black', alpha=0.65, ec='none'))
        x1, y1, x2, y2 = b.bbox
        axes[4].add_patch(mpatches.FancyBboxPatch(
            (x1, y1), (x2-x1), (y2-y1), boxstyle='square,pad=0',
            lw=1.5, edgecolor='yellow', facecolor='none'))
    if neighborhoods:
        for nb in neighborhoods:
            x1, y1, x2, y2 = nb.bbox
            clr = {'high': '#ff4444', 'medium': '#ffaa00', 'low': '#44ff44'}[nb.density_class]
            axes[4].add_patch(mpatches.Rectangle(
                (x1, y1), x2-x1, y2-y1, linewidth=2, edgecolor=clr,
                facecolor='none', linestyle='--', alpha=0.7))
            axes[4].text(x1+5, y1+15, nb.grid_id, color=clr, fontsize=8,
                         fontweight='bold', bbox=dict(fc='black', alpha=0.5, ec='none', pad=1))
    axes[4].set_title(f'Building IDs -- {n_b} registered', color='white', fontsize=13, pad=8)

    ax5 = axes[5]
    ax5.set_facecolor('#161b22')
    ax5.axis('off')
    ax5.set_xlim(0, 1); ax5.set_ylim(0, 1)
    def T(x, y, s, sz=11, col='white', fw='normal', ha='center'):
        ax5.text(x, y, s, transform=ax5.transAxes, fontsize=sz, color=col,
                 ha=ha, va='center', fontweight=fw)
    avg_a = int(np.mean([b.area_pixels for b in buildings])) if buildings else 0
    avg_c = float(np.mean([b.confidence for b in buildings])) if buildings else 0
    T(0.5, 0.95, 'Scene Statistics', 15, 'white', 'bold')
    draw_hline(ax5, 0.89)
    info = [('Buildings detected', f'{n_b}', '#FFB347'),
            ('Vegetation segments', f'{n_v}', '#32BE46'),
            ('Road segments', f'{n_r}', '#888'),
            ('Water segments', f'{n_w}', '#3791DC'),
            ('Image resolution', f'{w} x {h} px', '#87CEEB'),
            ('Avg building area', f'{avg_a:,} px', '#87CEEB'),
            ('Avg CLIP confidence', f'{avg_c:.3f}', '#FFB347')]
    for i, (k, v, col) in enumerate(info):
        yp = 0.82 - i * 0.07
        T(0.05, yp, k, 10, '#AAA', ha='left')
        T(0.95, yp, v, 10, col, ha='right')
    draw_hline(ax5, 0.30)
    T(0.5, 0.26, 'Land Use Breakdown', 12, 'white', 'bold')
    for pv, col, label, yp in [
        (pct['open'], '#5A6EC8', f"Open/Road {pct['open']:.1f}%", 0.19),
        (pct['veg'], '#32BE46', f"Vegetation {pct['veg']:.1f}%", 0.11),
        (pct['built'], '#D23737', f"Built-up {pct['built']:.1f}%", 0.03)]:
        ax5.barh(yp, 0.80*(pv/100), height=0.05, left=0.10, color=col,
                 alpha=0.85, transform=ax5.transAxes)
        ax5.barh(yp, 0.80, height=0.05, left=0.10, color=col,
                 alpha=0.15, transform=ax5.transAxes)
        T(0.91, yp, label, 9, col, ha='right')
    T(0.5, 0.002, 'UrbanInsight  |  SAM Auto + CLIP Classifier', 8, '#555')
    for ax in axes[:5]:
        ax.set_facecolor('#0d1117')
        for sp in ax.spines.values(): sp.set_edgecolor('#30363d'); sp.set_linewidth(1.2)
        ax.tick_params(colors='#555', labelsize=7)
    if save_path:
        plt.savefig(save_path, dpi=140, bbox_inches='tight',
                    facecolor='#0d1117', edgecolor='none')
        print(f'Saved: {save_path}')
    plt.show()

print("Visualization ready!")
```

```python
# ================================================================
# FULL PIPELINE: SAM Auto + CLIP Classification
# ================================================================
# SAM Auto segments the image (works great for aerial imagery)
# CLIP classifies each segment semantically (no color rules!)
# ================================================================

import time, json, numpy as np
from PIL import Image
from pathlib import Path

def run_pipeline(image_input, points_per_side=48,
                 min_bld_area=200, neighborhood_grid=(3, 3),
                 save_name='result'):
    t0 = time.time()

    if isinstance(image_input, (str, Path)):
        img = np.array(Image.open(str(image_input)).convert('RGB'))
    else:
        img = np.array(image_input.convert('RGB'))

    h, w = img.shape[:2]
    print(f"Image: {w} x {h} px")

    MAX_DIM = 1200
    scale = 1.0
    if max(h, w) > MAX_DIM:
        scale = MAX_DIM / max(h, w)
        nh, nw = int(h * scale), int(w * scale)
        img = np.array(Image.fromarray(img).resize((nw, nh), Image.LANCZOS))
        h, w = nh, nw
        print(f"Resized: {nw} x {nh}")

    adj_min = max(50, int(min_bld_area * scale**2))

    # Step 1: SAM Auto
    print(f"\n[1/6] SAM Auto segmentation (points={points_per_side})...")
    sam_masks = run_sam_auto(img, points_per_side=points_per_side,
                             pred_iou_thresh=0.76, stability_thresh=0.80,
                             min_area=max(50, adj_min // 4))
    print(f"  Raw segments: {len(sam_masks)}")

    # Step 2: CLIP classification
    print(f"\n[2/6] CLIP classification...")
    classified = classify_segments_batch(img, sam_masks, batch_size=16)

    # Step 3: Post-process buildings
    print(f"\n[3/6] Post-processing buildings...")
    raw_masks = [s['mask'] for s in classified['buildings']]
    cl_masks = clean_masks(raw_masks, opening=3, closing=7,
                           min_area=max(50, adj_min // 3))
    cl_masks = merge_overlapping(cl_masks, iou_threshold=0.3)
    print(f"  {len(raw_masks)} -> {len(cl_masks)} after cleaning")

    classified['buildings'] = []
    for m in cl_masks:
        sh = get_shape_info(m)
        classified['buildings'].append({
            'mask': m, 'area': int(m.sum()), 'quality': 1.0,
            'clip_score': 1.0, 'bbox': sh['bbox'],
            'centroid': sh['centroid'], 'shape': sh,
        })

    # Step 4: Land use map
    print(f"\n[4/6] Building land use map...")
    land_map = build_land_use_map(img, classified)
    total = land_map.size
    print(f"  Open: {np.sum(land_map==0)/total*100:.1f}%  "
          f"Veg: {np.sum(land_map==1)/total*100:.1f}%  "
          f"Built: {np.sum(land_map==2)/total*100:.1f}%  "
          f"Water: {np.sum(land_map==3)/total*100:.1f}%")

    # Step 5: Register
    print(f"\n[5/6] Registering buildings...")
    buildings = register_buildings(classified['buildings'])
    print(f"  Registered: {len(buildings)} buildings")

    # Step 6: Neighborhoods
    print(f"\n[6/6] Neighborhood analysis...")
    grid_r, grid_c = neighborhood_grid
    neighborhoods = analyze_neighborhoods(land_map, buildings,
                                          grid_rows=grid_r, grid_cols=grid_c)
    print_neighborhood_report(neighborhoods)

    elapsed = time.time() - t0
    print(f"\nTotal time: {elapsed:.1f}s")

    save_path = str(DIRS['vis'] / f'{save_name}.png')
    visualize_results(img, classified, buildings, land_map,
                      neighborhoods=neighborhoods, save_path=save_path)

    # Save JSONs
    reg = [{'id': b.building_id, 'bbox': b.bbox, 'area_px': b.area_pixels,
            'confidence': round(b.confidence, 4), 'centroid': b.centroid,
            'compactness': round(b.compactness, 4),
            'rectangularity': round(b.rectangularity, 4),
            'mask_file': f'{b.building_id}.png'} for b in buildings]
    with open(DIRS['results'] / f'{save_name}_buildings.json', 'w') as f:
        json.dump(reg, f, indent=2)

    nb_data = [{'grid_id': nb.grid_id, 'bbox': nb.bbox,
                'n_buildings': nb.n_buildings, 'building_ids': nb.building_ids,
                'land_use_pct': nb.land_use_pct,
                'density_class': nb.density_class} for nb in neighborhoods]
    with open(DIRS['results'] / f'{save_name}_neighborhoods.json', 'w') as f:
        json.dump(nb_data, f, indent=2)

    stats = {'pipeline': 'SAM Auto + CLIP', 'n_buildings': len(buildings),
             'n_vegetation': len(classified.get('vegetation', [])),
             'n_road': len(classified.get('road', [])),
             'n_water': len(classified.get('water', [])),
             'land_use': {
                 'open_road_pct': round(np.sum(land_map==0)/total*100, 2),
                 'veg_pct': round(np.sum(land_map==1)/total*100, 2),
                 'built_pct': round(np.sum(land_map==2)/total*100, 2),
                 'water_pct': round(np.sum(land_map==3)/total*100, 2)},
             'neighborhoods': nb_data,
             'elapsed_s': round(elapsed, 2)}
    with open(DIRS['results'] / f'{save_name}_stats.json', 'w') as f:
        json.dump(stats, f, indent=2)

    print(f"\nFiles saved to {DIRS['results']}")
    return {'image': img, 'buildings': buildings, 'classified': classified,
            'land_use_map': land_map, 'neighborhoods': neighborhoods}

print("Pipeline ready: SAM Auto + CLIP")
print("Usage: results = run_pipeline(your_image)")
```
