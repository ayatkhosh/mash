```python
# ================================================================
# IMPROVED AERIAL IMAGE CLASSIFIER  (v5 - better vegetation/water/open)
# ================================================================
#
# Key improvements over v4:
#   • Vegetation checked FIRST (before water) to avoid blue-green confusion
#   • Added ExG (Excess Green), VARI, and ExGR indices for robust vegetation
#   • Water detection requires both blue/cyan hue AND saturation (not just dark)
#   • Open land rule expanded to cover sandy/pale and dry-grass areas
#   • All thresholds documented inline for easy tuning
# ================================================================

import cv2, numpy as np
from scipy import ndimage


def classify_aerial_segment(image_rgb: np.ndarray,
                              mask: np.ndarray) -> str:
    """
    Classify an aerial segment using HSV + RGB + vegetation indices.

    Indices used
    ────────────
    ExG  = 2·G − R − B          (Excess Green; >15 → strong vegetation)
    VARI = (G−R)/(G+R−B)        (VARI; >0.05 → vegetation likely)
    greenness = G − 0.5·(R+B)   (simple green dominance)

    Priority order (highest→lowest):
      vegetation → water → road → buildings → open_land → unknown

    Returns one of:
      buildings | vegetation | road | open_land | water | unknown
    """
    pixels = image_rgb[mask]
    if len(pixels) < 20:
        return 'unknown'

    # ── RGB means ────────────────────────────────────────────
    R = float(pixels[:, 0].mean())
    G = float(pixels[:, 1].mean())
    B = float(pixels[:, 2].mean())

    # ── HSV stats ────────────────────────────────────────────
    bgr = pixels[:, ::-1].astype(np.uint8)
    hsv = cv2.cvtColor(bgr.reshape(-1, 1, 3), cv2.COLOR_BGR2HSV).reshape(-1, 3)
    H_mean = float(hsv[:, 0].mean())   # 0–180
    S_mean = float(hsv[:, 1].mean())   # 0–255
    V_mean = float(hsv[:, 2].mean())   # 0–255
    V_std  = float(hsv[:, 2].std())

    # ── Vegetation indices ────────────────────────────────────
    # ExG: positive = more green than average; >15 = clearly vegetation
    ExG = 2.0 * G - R - B
    # VARI: scale-invariant; >0.05 = vegetation likely
    vari_denom = G + R - B
    VARI = (G - R) / vari_denom if abs(vari_denom) > 1.0 else 0.0
    # Classic greenness: G clearly above average of R and B
    greenness = G - 0.5 * (R + B)

    # ── Rule 1: Vegetation ─────────────────────────────────── (checked FIRST)
    # Green/yellow-green hue (OpenCV H: 25–95 covers lime→green→teal)
    veg_hue      = 25 <= H_mean <= 92          # green/yellow-green hue range
    veg_sat      = S_mean >= 18                # must have some color
    veg_exg      = ExG >= 15                   # strong excess green
    veg_vari     = VARI >= 0.05                # VARI confirms vegetation
    veg_green    = greenness >= 6              # G clearly dominant

    if veg_hue and veg_sat and (veg_exg or veg_vari or veg_green):
        return 'vegetation'
    # Second pass: very strong ExG even outside the typical hue window
    if ExG >= 28 and S_mean >= 15 and greenness >= 10:
        return 'vegetation'

    # ── Rule 2: Water ─────────────────────────────────────────
    # Must have cyan/blue hue AND meaningful saturation (avoids shadow/road)
    # Water: H roughly 95–150 (blue-cyan), S >= 40, not pure black
    water_cyan   = 95 <= H_mean <= 150         # cyan→blue hue (OpenCV H)
    water_sat    = S_mean >= 38                # actual color, not just gray
    water_bright = V_mean >= 35                # not pure black (shadow)

    if water_cyan and water_sat and water_bright:
        return 'water'

    # Blue channel dominance (clear blue/indigo water under harsh light)
    # Require brightness > 40 to avoid confusing with dark asphalt
    if B > R + 14 and B > G + 6 and V_mean > 42 and S_mean >= 28:
        return 'water'

    # ── Rule 3: Road / Asphalt ───────────────────────────────
    # Dark gray, very uniform, almost no saturation
    if S_mean < 28 and V_mean < 95 and V_std < 28:
        return 'road'

    # ── Rule 4: Buildings ────────────────────────────────────
    # White/light metal rooftops
    if V_mean > 185 and S_mean < 45:
        return 'buildings'
    # Gray concrete rooftops (most common in aerial imagery)
    if S_mean < 55 and 88 <= V_mean <= 190 and V_std > 8:
        return 'buildings'
    # Warm-toned rooftops (terracotta, beige, brown)
    if 8 <= H_mean <= 32 and 18 <= S_mean <= 120 and V_mean >= 80:
        return 'buildings'
    # Dark rooftops (tar, dark tiles, felt)
    if S_mean < 42 and 45 <= V_mean <= 92:
        return 'buildings'

    # ── Rule 5: Open Land ─────────────────────────────────────
    # Bare soil/sand (warm, bright, low-medium saturation)
    open_warm      = 10 <= H_mean <= 38 and S_mean < 80 and V_mean > 100
    # Sandy/pale surface (light, pale, nearly achromatic)
    open_sandy     = S_mean < 45 and V_mean > 145
    # Dry grass / fallow field (yellow-green hue, low ExG, not dark)
    open_dry_grass = (40 <= H_mean <= 65 and S_mean < 65
                      and V_mean > 90 and ExG < 8)

    if open_warm or open_sandy or open_dry_grass:
        return 'open_land'

    # ── Fallback: non-green, not dark → likely building ───────
    if greenness < 3 and V_mean > 68:
        return 'buildings'

    return 'open_land'


def get_shape_info(mask: np.ndarray) -> dict:
    """Compute bounding box, centroid, aspect ratio, and area of a mask."""
    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)
    if not rows.any():
        return {'aspect_ratio': 1, 'rectangularity': 0, 'area': 0,
                'bbox': [0, 0, 0, 0], 'centroid': (0, 0)}
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    bh   = max(rmax - rmin + 1, 1)
    bw   = max(cmax - cmin + 1, 1)
    area = int(mask.sum())
    cy, cx = ndimage.center_of_mass(mask)
    return {
        'aspect_ratio'  : max(bh, bw) / min(bh, bw),
        'rectangularity': area / (bh * bw),
        'area'          : area,
        'bbox'          : [int(cmin), int(rmin), int(cmax), int(rmax)],
        'centroid'      : (int(cx), int(cy)),
    }


print("Classifier v5 ready!")
print("  Indices: ExG, VARI, greenness")
print("  Priority: vegetation → water → road → buildings → open_land")
```

```python
# ================================================================
# SAM AUTO + CLASSIFICATION ENGINE  (v5 - improved land-use map)
# ================================================================
#
# Key improvements:
#   • build_land_use_map uses ExG+VARI for pixel-level vegetation
#   • Water pixel map now uses B-channel dominance + cyan hue
#   • Priority ordering: buildings > water > vegetation > open/road
# ================================================================

from segment_anything import SamAutomaticMaskGenerator
import torch, numpy as np, cv2


def run_sam_auto(image_np: np.ndarray,
                 points_per_side: int = 48,
                 pred_iou_thresh: float = 0.80,
                 stability_thresh: float = 0.85,
                 min_area: int = 150) -> list:
    """Run SAM Automatic Mask Generator. Returns masks sorted by area (largest first)."""
    gen = SamAutomaticMaskGenerator(
        model                          = sam_model,
        points_per_side                = points_per_side,
        pred_iou_thresh                = pred_iou_thresh,
        stability_score_thresh         = stability_thresh,
        min_mask_region_area           = min_area,
        crop_n_layers                  = 1,
        crop_n_points_downscale_factor = 2,
    )
    with torch.no_grad():
        masks = gen.generate(image_np)
    return sorted(masks, key=lambda x: x['area'], reverse=True)


def classify_all_segments(image_np: np.ndarray,
                           sam_masks: list,
                           min_bld_area: int = 400,
                           max_bld_ratio: float = 0.20,
                           max_aspect: float = 14.0) -> dict:
    """
    Classify every SAM segment and return categorised lists.

    Output dict keys:  buildings | vegetation | road | open_land | water | unknown
    Each entry is a dict with: mask, area, quality, bbox, centroid, shape.
    """
    h, w  = image_np.shape[:2]
    total = h * w

    out = {c: [] for c in ['buildings', 'vegetation', 'road',
                             'open_land', 'water', 'unknown']}

    for m in sam_masks:
        mask = m['segmentation']
        area = m['area']
        iou  = m['predicted_iou']

        shape = get_shape_info(mask)
        if shape['area'] < 30:
            continue

        cat = classify_aerial_segment(image_np, mask)

        # ── Validate buildings (size & shape sanity checks) ──
        if cat == 'buildings':
            if area / total > max_bld_ratio:
                cat = 'open_land'           # too large to be a single building
            elif shape['aspect_ratio'] > max_aspect:
                cat = 'road'                # very elongated → road/fence
            elif shape['area'] < min_bld_area:
                cat = 'unknown'             # too small

        if cat not in out:
            cat = 'unknown'

        out[cat].append({
            'mask'    : mask,
            'area'    : area,
            'quality' : iou,
            'bbox'    : shape['bbox'],
            'centroid': shape['centroid'],
            'shape'   : shape,
        })

    return out


def build_land_use_map(image_np: np.ndarray,
                        classified: dict) -> np.ndarray:
    """
    Build a pixel-level land use map.

    Label values
    ────────────
      0 = open / road
      1 = vegetation
      2 = building
      3 = water

    Strategy
    ────────
    1. Pixel-level base layer using improved vegetation & water indices
    2. Overlay SAM segments (more precise boundaries) in priority order
       Priority: buildings > water > vegetation  (lower wins ↓)
    """
    h, w = image_np.shape[:2]

    # ── Pixel-level features ──────────────────────────────────
    hsv = cv2.cvtColor(image_np, cv2.COLOR_RGB2HSV)
    H   = hsv[:, :, 0].astype(float)   # 0–180
    S   = hsv[:, :, 1].astype(float)   # 0–255
    V   = hsv[:, :, 2].astype(float)   # 0–255

    R  = image_np[:, :, 0].astype(float)
    G  = image_np[:, :, 1].astype(float)
    Bc = image_np[:, :, 2].astype(float)

    # Vegetation indices (pixel-level)
    ExG = 2.0 * G - R - Bc                         # Excess Green
    vari_denom = G + R - Bc
    # Avoid division by zero; VARI meaningful only where denominator != 0
    VARI = np.where(np.abs(vari_denom) > 1.0,
                    (G - R) / vari_denom, 0.0)
    greenness = G - 0.5 * (R + Bc)

    lmap = np.zeros((h, w), dtype=np.uint8)           # default = 0 (open/road)

    # Pixel-level vegetation (label 1)
    # Require green hue AND at least one of the green indices
    veg_mask = (
        (H >= 25) & (H <= 92) & (S >= 18) &
        ((ExG >= 15) | (VARI >= 0.05) | (greenness >= 6))
    )
    lmap[veg_mask] = 1

    # Pixel-level water (label 3)
    # Cyan/blue hue + saturation + not too dark  (avoids road/shadow confusion)
    water_hue_mask = (H >= 95) & (H <= 150) & (S >= 38) & (V >= 35)
    water_blue_dom = (Bc > R + 14) & (Bc > G + 6) & (V > 42) & (S >= 28)
    lmap[water_hue_mask | water_blue_dom] = 3

    # ── SAM segment overlay (priority: veg < water < buildings) ──
    # Vegetation first (lowest priority) …
    for seg in classified.get('vegetation', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 1

    # … then water (overwrites misclassified vegetation) …
    for seg in classified.get('water', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 3

    # … buildings last (highest priority — always on top)
    for seg in classified.get('buildings', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 2

    return lmap


print("SAM + Classification engine v5 ready!")
print("  build_land_use_map: ExG/VARI vegetation | blue-channel water")
print("  Priority: buildings > water > vegetation > open/road")
print("  Dict keys: buildings / vegetation / road / open_land / water / unknown")
```

```python
# ================================================================
# POST-PROCESSING & BUILDING REGISTRY
# ================================================================

import cv2, numpy as np, uuid
from dataclasses import dataclass
from typing import List, Tuple
from scipy import ndimage
from PIL import Image


@dataclass
class Building:
    building_id : str
    bbox        : List[int]      # [x1, y1, x2, y2]
    area_pixels : int
    confidence  : float
    mask        : np.ndarray
    centroid    : Tuple[int,int]


def clean_masks(masks: List[np.ndarray],
                opening: int = 3,
                closing: int = 7,
                min_area: int = 200) -> List[np.ndarray]:
    '''
    Morphological cleaning:
    1. Opening  -> remove small noise
    2. Closing  -> fill holes in rooftops
    3. Filter   -> remove still-too-small regions
    '''
    out = []
    for m in masks:
        u  = m.astype(np.uint8) * 255
        k1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(opening,opening))
        k2 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(closing,closing))
        u  = cv2.morphologyEx(u, cv2.MORPH_OPEN,  k1)
        u  = cv2.morphologyEx(u, cv2.MORPH_CLOSE, k2)
        n,labels,stats,_ = cv2.connectedComponentsWithStats(u,8)
        clean = np.zeros_like(u)
        for lid in range(1,n):
            if stats[lid, cv2.CC_STAT_AREA] >= min_area:
                clean[labels==lid] = 255
        result = clean > 0
        if result.sum() >= min_area:
            out.append(result)
    return out


def register_buildings(classified_blds: List[dict],
                        image_path: str = '') -> List[Building]:
    '''Assign unique UUID to each building and save its mask.'''
    blds = []
    for seg in classified_blds:
        mask = seg['mask']
        rows = np.any(mask, axis=1)
        cols = np.any(mask, axis=0)
        if not rows.any(): continue
        rmin,rmax = np.where(rows)[0][[0,-1]]
        cmin,cmax = np.where(cols)[0][[0,-1]]
        cy,cx     = ndimage.center_of_mass(mask)
        bid       = f'BLD-{str(uuid.uuid4())[:8].upper()}'
        Image.fromarray(mask.astype(np.uint8)*255).save(
            DIRS['masks'] / f'{bid}.png')
        blds.append(Building(
            building_id = bid,
            bbox        = [int(cmin),int(rmin),int(cmax),int(rmax)],
            area_pixels = int(mask.sum()),
            confidence  = float(seg['quality']),
            mask        = mask,
            centroid    = (int(cx),int(cy)),
        ))
    blds.sort(key=lambda b: b.area_pixels, reverse=True)
    return blds


print("Post-processing & registry ready!")
```

```python
# ================================================================
# VISUALIZATION ENGINE (axhline bug fixed)
# ================================================================

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
plt.rcParams['font.family'] = 'DejaVu Sans'

CAT_COLORS = {
    'buildings' : (220, 70,  50),
    'vegetation': ( 50,195,  70),
    'road'      : (110,110, 110),
    'open_land' : (180,155,  90),
    'water'     : ( 60,140, 220),
    'unknown'   : (170,170, 170),
}
CAT_NAMES = {
    'buildings' :'Buildings',
    'vegetation':'Vegetation',
    'road'      :'Road',
    'open_land' :'Open Land',
    'water'     :'Water',
    'unknown'   :'Other',
}


def colored_overlay(image_np: np.ndarray,
                     masks: list,
                     alpha: float = 0.62) -> np.ndarray:
    ov   = image_np.copy().astype(float)
    cmap = plt.get_cmap('tab20')
    for i,m in enumerate(masks):
        if m.shape != image_np.shape[:2]: continue
        c = np.array(cmap(i % 20)[:3]) * 255
        for ch in range(3):
            ov[:,:,ch][m] = ov[:,:,ch][m]*(1-alpha) + c[ch]*alpha
    return ov.astype(np.uint8)


def draw_hline(ax, y, xmin=0.05, xmax=0.95, color='#444', lw=1):
    '''Fixed version of axhline that avoids the transform conflict.'''
    # Use plot in axis coordinates instead of axhline
    ax.plot([xmin, xmax], [y, y],
            color=color, linewidth=lw,
            transform=ax.transAxes,
            clip_on=False, solid_capstyle='butt')


def visualize_results(image_np, classified, buildings, land_map,
                       title='UrbanInsight P1 FINAL', save_path=None):
    '''
    6-panel professional layout:
    [Original] [SAM All Classified] [Building Masks]
    [Land Use] [Building IDs]       [Statistics]
    '''
    fig = plt.figure(figsize=(26,17))
    fig.patch.set_facecolor('#0d1117')

    n_b = len(buildings)
    n_v = len(classified.get('vegetation',[]))
    n_r = len(classified.get('road',[]))
    total_seg = sum(len(classified.get(c,[])) for c in CAT_COLORS)

    fig.suptitle(
        f'{title}  |  Buildings: {n_b}  '
        f'Vegetation: {n_v}  Road: {n_r}  Total segments: {total_seg}',
        fontsize=15, fontweight='bold', color='white', y=0.99
    )

    gs   = fig.add_gridspec(2,3, hspace=0.28, wspace=0.14,
                            left=0.02, right=0.98,
                            top=0.94, bottom=0.02)
    axes = [fig.add_subplot(gs[r,c]) for r in range(2) for c in range(3)]
    h,w  = image_np.shape[:2]

    # ── [0] Original ──────────────────────────────────────────
    axes[0].imshow(image_np)
    axes[0].set_title('Original Image', color='white', fontsize=13, pad=8)

    # ── [1] All SAM segments colored by class ─────────────────
    ov1      = image_np.copy().astype(float)
    leg1     = []
    for cat,(r,g,b) in CAT_COLORS.items():
        segs = classified.get(cat,[])
        if not segs: continue
        c = np.array([r,g,b], dtype=float)
        for seg in segs:
            m = seg['mask']
            if m.shape != (h,w): continue
            for ch in range(3):
                ov1[:,:,ch][m] = ov1[:,:,ch][m]*0.25 + c[ch]*0.75
        leg1.append(mpatches.Patch(
            color=np.array([r,g,b])/255,
            label=f'{CAT_NAMES[cat]} ({len(segs)})'))
    axes[1].imshow(ov1.astype(np.uint8))
    if leg1:
        axes[1].legend(handles=leg1, loc='upper right', fontsize=8.5,
                       framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[1].set_title(f'SAM Classification -- {total_seg} segments',
                      color='white', fontsize=13, pad=8)

    # ── [2] Building masks ────────────────────────────────────
    bld_masks = [b.mask for b in buildings]
    axes[2].imshow(colored_overlay(image_np, bld_masks, 0.70))
    axes[2].set_title(f'Building Masks -- {n_b} buildings',
                      color='white', fontsize=13, pad=8)

    # ── [3] Land Use Map ──────────────────────────────────────
    lrgb = np.zeros((h,w,3), dtype=np.uint8)
    lrgb[land_map==0] = [ 90,110,200]
    lrgb[land_map==1] = [ 50,190, 70]
    lrgb[land_map==2] = [210, 55, 55]
    lrgb[land_map==3] = [ 55,145,220]
    total_px = land_map.size
    pct = {k: np.sum(land_map==v)/total_px*100
           for k,v in {'open':0,'veg':1,'built':2,'water':3}.items()}
    axes[3].imshow(lrgb)
    lu_leg = [
        mpatches.Patch(color='#5A6EC8',
                       label=f"Open/Road  {pct['open']:.1f}%"),
        mpatches.Patch(color='#32BE46',
                       label=f"Vegetation {pct['veg']:.1f}%"),
        mpatches.Patch(color='#D23737',
                       label=f"Built-up   {pct['built']:.1f}%"),
        mpatches.Patch(color='#3791DC',
                       label=f"Water      {pct['water']:.1f}%"),
    ]
    axes[3].legend(handles=lu_leg, loc='upper right', fontsize=9,
                   framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[3].set_title('Land Use Map (FR-4)',
                      color='white', fontsize=13, pad=8)

    # ── [4] Building IDs ──────────────────────────────────────
    axes[4].imshow(colored_overlay(image_np, bld_masks, 0.55))
    for b in buildings:
        cx,cy = b.centroid
        fs    = max(5, min(9, int(b.area_pixels**0.38 * 0.30)))
        axes[4].text(
            cx, cy, b.building_id[-6:],
            color='white', fontsize=fs,
            ha='center', va='center', fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.15',
                      fc='black', alpha=0.65, ec='none'))
        x1,y1,x2,y2 = b.bbox
        axes[4].add_patch(mpatches.FancyBboxPatch(
            (x1,y1),(x2-x1),(y2-y1),
            boxstyle='square,pad=0', lw=1.5,
            edgecolor='yellow', facecolor='none'))
    axes[4].set_title(f'Building IDs -- {n_b} registered',
                      color='white', fontsize=13, pad=8)

    # ── [5] Statistics panel ──────────────────────────────────
    ax5 = axes[5]
    ax5.set_facecolor('#161b22')
    ax5.axis('off')
    ax5.set_xlim(0,1)
    ax5.set_ylim(0,1)

    def T(x, y, s, sz=11, col='white', fw='normal', ha='center'):
        ax5.text(x, y, s, transform=ax5.transAxes,
                 fontsize=sz, color=col,
                 ha=ha, va='center', fontweight=fw)

    avg_a = int(np.mean([b.area_pixels for b in buildings])) if buildings else 0
    avg_c = float(np.mean([b.confidence for b in buildings])) if buildings else 0

    T(0.5, 0.95, 'Scene Statistics', 15, 'white', 'bold')
    draw_hline(ax5, 0.89)   # <-- FIXED: no transform conflict

    info_rows = [
        ('Buildings detected', f'{n_b}',          '#FFB347'),
        ('Vegetation segments',f'{n_v}',           '#32BE46'),
        ('Water segments',    f'{len(classified.get("water",[]))}', '#3791DC'),
        ('Image resolution',   f'{w} x {h} px',   '#87CEEB'),
        ('Avg building area',  f'{avg_a:,} px',   '#87CEEB'),
        ('Avg SAM quality',    f'{avg_c:.3f}',    '#87CEEB'),
    ]
    for i,(k,v,col) in enumerate(info_rows):
        yp = 0.81 - i*0.08
        T(0.05, yp, k, 10, '#AAA', ha='left')
        T(0.95, yp, v, 10, col,   ha='right')

    draw_hline(ax5, 0.37)   # <-- FIXED
    T(0.5, 0.33, 'Land Use Breakdown', 12, 'white', 'bold')

    bar_items = [
        (pct['built'], '#D23737', f"Built-up   {pct['built']:.1f}%", 0.27),
        (pct['veg'],   '#32BE46', f"Vegetation {pct['veg']:.1f}%",   0.19),
        (pct['water'], '#3791DC', f"Water      {pct['water']:.1f}%", 0.11),
        (pct['open'],  '#5A6EC8', f"Open/Road  {pct['open']:.1f}%",  0.03),
    ]
    for pv, col, label, yp in bar_items:
        ax5.barh(yp, 0.80*(pv/100), height=0.055,
                 left=0.10, color=col, alpha=0.85,
                 transform=ax5.transAxes)
        ax5.barh(yp, 0.80, height=0.055,
                 left=0.10, color=col, alpha=0.15,
                 transform=ax5.transAxes)
        T(0.91, yp, label, 9.5, col, ha='right')

    draw_hline(ax5, 0.008, color='#333', lw=0.8)  # <-- FIXED
    T(0.5, 0.002,
      'UrbanInsight FINAL  |  SAM + Aerial Color Classifier',
      8, '#555')

    # General styling
    for ax in axes[:5]:
        ax.set_facecolor('#0d1117')
        for sp in ax.spines.values():
            sp.set_edgecolor('#30363d')
            sp.set_linewidth(1.2)
        ax.tick_params(colors='#555', labelsize=7)

    if save_path:
        plt.savefig(save_path, dpi=140, bbox_inches='tight',
                    facecolor='#0d1117', edgecolor='none')
        print(f'Saved: {save_path}')
    plt.show()


print("Visualization ready!")
print("  draw_hline() - replaces axhline to avoid transform conflict")
```

```python
# ================================================================
# FULL PIPELINE  (v5 - enhanced outputs + water stats)
# ================================================================
#
# New outputs:
#   <name>_buildings.json  : id, bbox, centroid, area_px, confidence
#   <name>_vegetation.json : id, bbox, centroid, area_px, quality
#   <name>_water.json      : id, bbox, centroid, area_px, quality
#   <name>_stats.json      : land-use percentages for ALL 4 categories
#   <name>_landuse.json    : full structured land-use registry
# ================================================================

import time, json, numpy as np
from PIL import Image
from pathlib import Path


def run_pipeline(image_input,
                 points_per_side: int = 48,
                 min_bld_area: int = 400,
                 save_name: str = 'result') -> dict:
    """
    Complete pipeline
    ─────────────────
    1. SAM Automatic mask generation
    2. Aerial colour+shape classification (v5 – better veg/water/open)
    3. Land use map (pixel-level, 4 classes)
    4. Morphological post-processing of building masks
    5. Building registry (UUIDs + bbox + centroid + area + confidence)
    6. Vegetation & water registries (consistent JSON schema)
    7. Professional 6-panel visualisation + JSON exports

    Parameters
    ──────────
    points_per_side : SAM grid density (32 = fast, 48 = balanced, 64 = best)
    min_bld_area    : minimum building size in pixels
    save_name       : filename prefix for all saved outputs
    """
    t0 = time.time()

    # ── Load image ───────────────────────────────────────────
    if isinstance(image_input, (str, Path)):
        img = np.array(Image.open(str(image_input)).convert('RGB'))
    else:
        img = np.array(image_input.convert('RGB'))

    h, w = img.shape[:2]
    print(f"Image: {w} × {h} px")

    # Resize if larger than 1500 px on any side (memory management)
    MAX_DIM = 1500
    scale   = 1.0
    if max(h, w) > MAX_DIM:
        scale      = MAX_DIM / max(h, w)
        nh, nw     = int(h * scale), int(w * scale)
        img        = np.array(Image.fromarray(img).resize((nw, nh), Image.LANCZOS))
        h, w       = nh, nw
        print(f"Resized: {nw} × {nh}")

    adj_min = max(80, int(min_bld_area * scale ** 2))
    total   = h * w

    # ── Step 1: SAM Auto ─────────────────────────────────────
    print(f"\n[1/5] SAM Automatic (points_per_side={points_per_side})...")
    sam_masks = run_sam_auto(
        img,
        points_per_side  = points_per_side,
        pred_iou_thresh  = 0.80,
        stability_thresh = 0.85,
        min_area         = max(50, adj_min // 4),
    )
    print(f"  Raw segments: {len(sam_masks)}")

    # ── Step 2: Classify ─────────────────────────────────────
    print("\n[2/5] Classifying segments...")
    classified = classify_all_segments(
        img, sam_masks,
        min_bld_area  = adj_min,
        max_bld_ratio = 0.20,
        max_aspect    = 14.0,
    )
    for cat, segs in classified.items():
        if segs:
            print(f"  {cat:<14}: {len(segs):>4}")

    # ── Step 3: Land use map ─────────────────────────────────
    print("\n[3/5] Land use map (ExG+VARI vegetation, blue-channel water)...")
    land_map = build_land_use_map(img, classified)

    pct_open  = np.sum(land_map == 0) / total * 100
    pct_veg   = np.sum(land_map == 1) / total * 100
    pct_built = np.sum(land_map == 2) / total * 100
    pct_water = np.sum(land_map == 3) / total * 100
    print(f"  Open:  {pct_open:.1f}%  "
          f"Veg: {pct_veg:.1f}%  "
          f"Built: {pct_built:.1f}%  "
          f"Water: {pct_water:.1f}%")

    # ── Step 4: Post-process building masks ──────────────────
    print("\n[4/5] Cleaning building masks...")
    raw_masks = [s['mask'] for s in classified['buildings']]
    cl_masks  = clean_masks(raw_masks,
                             opening  = 3,
                             closing  = 7,
                             min_area = max(50, adj_min // 3))
    print(f"  {len(raw_masks)} → {len(cl_masks)} after morphological cleaning")

    # Rebuild classified with cleaned masks
    classified['buildings'] = []
    for m in cl_masks:
        sh = get_shape_info(m)
        classified['buildings'].append({
            'mask': m, 'area': int(m.sum()), 'quality': 1.0,
            'bbox': sh['bbox'], 'centroid': sh['centroid'], 'shape': sh,
        })

    # ── Step 5: Building registry ─────────────────────────────
    print("\n[5/5] Registering buildings...")
    buildings = register_buildings(classified['buildings'])
    print(f"  Registered: {len(buildings)} buildings")

    elapsed = time.time() - t0
    print(f"\nTotal time: {elapsed:.1f}s")

    # ── Visualise ────────────────────────────────────────────
    save_path = str(DIRS['vis'] / f'{save_name}.png')
    visualize_results(img, classified, buildings, land_map,
                      title='UrbanInsight v5',
                      save_path=save_path)

    # ── Export: buildings JSON ────────────────────────────────
    # Schema: id, bbox [x1,y1,x2,y2], centroid (cx,cy),
    #         area_px, confidence
    building_reg = [
        {
            'id'        : b.building_id,
            'bbox'      : b.bbox,
            'centroid'  : list(b.centroid),
            'area_px'   : b.area_pixels,
            'confidence': round(b.confidence, 4),
        }
        for b in buildings
    ]
    with open(DIRS['results'] / f'{save_name}_buildings.json', 'w') as f:
        json.dump(building_reg, f, indent=2)

    # ── Export: vegetation JSON (same schema, quality from SAM) ──
    veg_reg = [
        {
            'id'     : f'VEG-{i+1:04d}',
            'bbox'   : seg['bbox'],
            'centroid': list(seg['centroid']),
            'area_px': int(seg['area']),
            'quality': round(float(seg['quality']), 4),
        }
        for i, seg in enumerate(classified.get('vegetation', []))
    ]
    with open(DIRS['results'] / f'{save_name}_vegetation.json', 'w') as f:
        json.dump(veg_reg, f, indent=2)

    # ── Export: water JSON ────────────────────────────────────
    water_reg = [
        {
            'id'     : f'WTR-{i+1:04d}',
            'bbox'   : seg['bbox'],
            'centroid': list(seg['centroid']),
            'area_px': int(seg['area']),
            'quality': round(float(seg['quality']), 4),
        }
        for i, seg in enumerate(classified.get('water', []))
    ]
    with open(DIRS['results'] / f'{save_name}_water.json', 'w') as f:
        json.dump(water_reg, f, indent=2)

    # ── Export: full land-use stats JSON ─────────────────────
    # Includes percentages for all 4 categories + segment counts
    stats = {
        'n_buildings'      : len(buildings),
        'n_vegetation_segs': len(classified.get('vegetation', [])),
        'n_water_segs'     : len(classified.get('water', [])),
        'land_use_percent' : {
            'built_up_pct'  : round(pct_built, 2),   # مناطق مبنية
            'vegetation_pct': round(pct_veg,   2),   # مناطق خضراء
            'open_land_pct' : round(pct_open,  2),   # مساحات مفتوحة
            'water_pct'     : round(pct_water, 2),   # مناطق مياه
        },
        'elapsed_s': round(elapsed, 2),
    }
    with open(DIRS['results'] / f'{save_name}_stats.json', 'w') as f:
        json.dump(stats, f, indent=2, ensure_ascii=False)

    # ── Export: consolidated land-use registry ────────────────
    landuse_registry = {
        'image_size'    : {'width': w, 'height': h},
        'land_use_percent': stats['land_use_percent'],
        'categories': {
            'built_up'  : building_reg,
            'vegetation': veg_reg,
            'water'     : water_reg,
        },
    }
    with open(DIRS['results'] / f'{save_name}_landuse.json', 'w') as f:
        json.dump(landuse_registry, f, indent=2, ensure_ascii=False)

    print(f"\nExported JSON files to {DIRS['results']}:")
    print(f"  {save_name}_buildings.json  ({len(building_reg)} buildings)")
    print(f"  {save_name}_vegetation.json ({len(veg_reg)} segments)")
    print(f"  {save_name}_water.json      ({len(water_reg)} segments)")
    print(f"  {save_name}_stats.json")
    print(f"  {save_name}_landuse.json    (consolidated)")

    return {
        'image'        : img,
        'buildings'    : buildings,
        'classified'   : classified,
        'land_use_map' : land_map,
        'stats'        : stats,
    }


print("Pipeline v5 ready!")
print()
print("Usage: results = run_pipeline(your_image_or_path)")
print()
print("Key parameters:")
print("  points_per_side=64  → more detections (slower)")
print("  min_bld_area=250    → detect smaller buildings")
print("  min_bld_area=600    → only larger structures")
```
