```python
# ================================================================
# CLIP-BASED AERIAL SEGMENT CLASSIFIER  (محسّن / Improved)
# ================================================================
# التحسينات:
#   1. Prompt Ensembles — عدة جمل لكل صنف بدل جملة واحدة
#   2. Confidence Threshold — unknown إذا الثقة أقل من الحد
#   3. Gray Background Masking — خلفية رمادية بدل التعتيم
#   4. Pre-CLIP Area Filter  — حذف القطع الصغيرة جداً
# ================================================================

import torch
import numpy as np
import cv2
from PIL import Image as PILImage
from scipy import ndimage

# -------------------------------------------------------------------
# Prompt Ensembles per category
# -------------------------------------------------------------------
CLIP_PROMPTS = {
    "buildings": [
        "aerial view of a building rooftop",
        "satellite photo of house roofs",
        "aerial image of residential buildings from above",
        "top view of urban building blocks",
        "aerial view of industrial or commercial building roof",
    ],
    "vegetation": [
        "aerial view of green trees and forest",
        "satellite image of vegetation and grass",
        "top view of park with grass and shrubs",
        "aerial view of farm fields and crops",
        "dense tree canopy from above",
    ],
    "road": [
        "aerial view of a road or highway",
        "satellite image of street pavement",
        "top view of asphalt road",
        "aerial view of parking lot or pathway",
        "satellite photo of road intersection",
    ],
    "water": [
        "aerial view of a river or lake",
        "satellite image of water body",
        "top view of swimming pool or pond",
        "aerial view of sea coastline",
        "satellite photo of flooded area",
    ],
    "open_land": [
        "aerial view of bare soil empty land",
        "satellite image of open field",
        "top view of sand or desert",
        "aerial view of construction site",
        "satellite photo of vacant lot",
    ],
}

CLIP_CATEGORIES = list(CLIP_PROMPTS.keys())

# Flat list of ALL prompts (for batch CLIP call)
ALL_PROMPTS = []
PROMPT_CAT_MAP = []   # which category each prompt belongs to
for cat, prompts in CLIP_PROMPTS.items():
    for p in prompts:
        ALL_PROMPTS.append(p)
        PROMPT_CAT_MAP.append(cat)

# Pre-compute text features once for speed
print("Pre-computing CLIP text features for ensemble prompts...")
_text_inputs = clip_processor(
    text=ALL_PROMPTS, return_tensors="pt", padding=True
).to(device)
with torch.no_grad():
    _text_features = clip_model.get_text_features(**_text_inputs)
    _text_features = _text_features / _text_features.norm(dim=-1, keepdim=True)
print(f"  Text features ready: {_text_features.shape}  ({len(ALL_PROMPTS)} prompts)")


# Configuration
CLIP_CONFIG = {
    "confidence_threshold": 0.22,   # below this -> unknown
    "min_crop_size": 20,            # minimum crop dimension (pixels)
    "crop_padding": 15,             # padding around bounding box
    "bg_gray_value": 128,           # gray value for masked background
    "min_area_px": 80,              # minimum segment area (pixels)
    "min_area_ratio": 0.00015,      # minimum area as fraction of image
}


def _make_crop(image_np, mask):
    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)
    if not rows.any():
        return None

    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]

    pad = CLIP_CONFIG["crop_padding"]
    h, w = image_np.shape[:2]
    rmin = max(0, rmin - pad)
    rmax = min(h - 1, rmax + pad)
    cmin = max(0, cmin - pad)
    cmax = min(w - 1, cmax + pad)

    crop = image_np[rmin:rmax+1, cmin:cmax+1].copy()
    mask_crop = mask[rmin:rmax+1, cmin:cmax+1]

    # Apply gray background to non-segment pixels
    g = CLIP_CONFIG["bg_gray_value"]
    for ch in range(3):
        crop[:, :, ch] = np.where(mask_crop, crop[:, :, ch], g)

    return crop


def _scores_from_image_features(img_feats):
    sim = (img_feats @ _text_features.T).cpu().numpy()  # (N, P)

    n = img_feats.shape[0]
    n_cats = len(CLIP_CATEGORIES)
    cat_scores = np.zeros((n, n_cats), dtype=np.float32)

    for p_idx, cat in enumerate(PROMPT_CAT_MAP):
        cat_idx = CLIP_CATEGORIES.index(cat)
        cat_scores[:, cat_idx] = np.maximum(cat_scores[:, cat_idx], sim[:, p_idx])

    # Softmax over categories
    e = np.exp(cat_scores - cat_scores.max(axis=1, keepdims=True))
    cat_scores = e / e.sum(axis=1, keepdims=True)
    return cat_scores


def classify_segments_batch(image_np, sam_masks, batch_size=16):
    h_img, w_img = image_np.shape[:2]
    total_px = h_img * w_img
    min_area_px = max(
        CLIP_CONFIG["min_area_px"],
        int(total_px * CLIP_CONFIG["min_area_ratio"])
    )
    min_crop = CLIP_CONFIG["min_crop_size"]
    conf_thresh = CLIP_CONFIG["confidence_threshold"]

    out = {c: [] for c in CLIP_CATEGORIES + ['unknown']}
    removed_by_area = 0

    crops = []
    indices = []

    for idx, m in enumerate(sam_masks):
        mask = m['segmentation']
        area = m['area']

        if area < min_area_px:
            removed_by_area += 1
            continue

        crop = _make_crop(image_np, mask)
        if crop is None or crop.shape[0] < min_crop or crop.shape[1] < min_crop:
            removed_by_area += 1
            continue

        crops.append(PILImage.fromarray(crop))
        indices.append(idx)

    print(f"  Classifying {len(crops)} segments with CLIP...")
    print(f"  (Removed by area filter: {removed_by_area})")

    # Process in batches
    for batch_start in range(0, len(crops), batch_size):
        batch_end = min(batch_start + batch_size, len(crops))
        batch_crops = crops[batch_start:batch_end]
        batch_indices = indices[batch_start:batch_end]

        img_inputs = clip_processor(
            images=batch_crops, return_tensors="pt", padding=True
        ).to(device)
        with torch.no_grad():
            img_feats = clip_model.get_image_features(**img_inputs)
            img_feats = img_feats / img_feats.norm(dim=-1, keepdim=True)

        cat_scores = _scores_from_image_features(img_feats)

        for i, idx in enumerate(batch_indices):
            m = sam_masks[idx]
            mask = m['segmentation']
            area = m['area']
            iou = m['predicted_iou']

            scores = cat_scores[i]
            best_idx = int(np.argmax(scores))
            confidence = float(scores[best_idx])

            if confidence < conf_thresh:
                cat = 'unknown'
            else:
                cat = CLIP_CATEGORIES[best_idx]

            shape = get_shape_info(mask)

            # Shape-based post-validation for buildings
            if cat == 'buildings':
                if area / total_px > 0.25:
                    cat = 'open_land'
                elif shape['aspect_ratio'] > 12:
                    cat = 'road'
                elif shape['compactness'] < 0.05:
                    cat = 'road'
                elif shape['area'] < min_area_px:
                    cat = 'unknown'

            out[cat].append({
                'mask': mask,
                'area': area,
                'quality': iou,
                'clip_score': confidence,
                'clip_scores': scores.tolist(),
                'bbox': shape['bbox'],
                'centroid': shape['centroid'],
                'shape': shape,
            })

        if (batch_end) % 64 == 0 or batch_end == len(crops):
            print(f"    Processed {batch_end}/{len(crops)}")

    print()
    for cat, segs in out.items():
        if segs:
            print(f"  {cat:<14}: {len(segs):>4}")
    print(f"  removed_by_area: {removed_by_area}")
    out['_stats'] = {'removed_by_area': removed_by_area}

    return out


def get_shape_info(mask):
    rows = np.any(mask, axis=1)
    cols = np.any(mask, axis=0)
    if not rows.any():
        return {'aspect_ratio': 1, 'rectangularity': 0, 'area': 0,
                'bbox': [0, 0, 0, 0], 'centroid': (0, 0), 'compactness': 0,
                'solidity': 0}
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    bh = max(rmax - rmin + 1, 1)
    bw = max(cmax - cmin + 1, 1)
    area = int(mask.sum())
    cy, cx = ndimage.center_of_mass(mask)
    contours, _ = cv2.findContours(
        mask.astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    perimeter = sum(cv2.arcLength(c, True) for c in contours) if contours else 1
    compactness = (4 * np.pi * area) / (perimeter**2 + 1e-6)
    hull_area = 0
    if contours:
        hull = cv2.convexHull(contours[0])
        hull_area = cv2.contourArea(hull)
    solidity = area / max(hull_area, 1)
    return {
        'aspect_ratio': max(bh, bw) / max(min(bh, bw), 1),
        'rectangularity': area / (bh * bw),
        'area': area,
        'bbox': [int(cmin), int(rmin), int(cmax), int(rmax)],
        'centroid': (int(cx), int(cy)),
        'compactness': float(min(compactness, 1.0)),
        'solidity': float(min(solidity, 1.0)),
    }


print("CLIP classifier ready (ensemble prompts + confidence thresholding)")
print(f"   Categories: {CLIP_CATEGORIES}")
print(f"   Total prompts: {len(ALL_PROMPTS)}")
print(f"   Confidence threshold: {CLIP_CONFIG['confidence_threshold']}")
```

```python
# ================================================================
# SAM AUTO-SEGMENTATION + LAND USE MAP  (محسّن / Improved)
# ================================================================
# التحسينات:
#   1. معاملات قابلة للضبط — Configurable SAM parameters
#   2. إغلاق وفتح مورفولوجي — Mask refinement (opening/closing)
# ================================================================

from segment_anything import SamAutomaticMaskGenerator

# -------------------------------------------------------------------
# SAM Configuration — عدّل هنا حسب صورتك
# -------------------------------------------------------------------
SAM_CONFIG = {
    "points_per_side": 48,           # عدد نقاط الشبكة (32/48/64/96)
    "pred_iou_thresh": 0.76,         # حد الثقة بالـ IoU (0.70-0.88)
    "stability_score_thresh": 0.80,  # حد الاستقرار (0.75-0.92)
    "crop_n_layers": 1,              # طبقات التقطيع (0/1/2)
    "crop_n_points_downscale_factor": 2,
    "min_mask_region_area": 100,     # أصغر منطقة (بكسل)
    "box_nms_thresh": 0.7,
}


def run_sam_auto(image_np,
                 points_per_side=None,
                 pred_iou_thresh=None,
                 stability_thresh=None,
                 min_area=None,
                 refine_masks=True):
    cfg = SAM_CONFIG.copy()
    if points_per_side is not None:  cfg["points_per_side"] = points_per_side
    if pred_iou_thresh  is not None: cfg["pred_iou_thresh"] = pred_iou_thresh
    if stability_thresh is not None: cfg["stability_score_thresh"] = stability_thresh
    if min_area         is not None: cfg["min_mask_region_area"] = min_area

    gen = SamAutomaticMaskGenerator(
        model=sam_model,
        points_per_side=cfg["points_per_side"],
        pred_iou_thresh=cfg["pred_iou_thresh"],
        stability_score_thresh=cfg["stability_score_thresh"],
        min_mask_region_area=cfg["min_mask_region_area"],
        crop_n_layers=cfg["crop_n_layers"],
        crop_n_points_downscale_factor=cfg["crop_n_points_downscale_factor"],
        box_nms_thresh=cfg["box_nms_thresh"],
    )
    with torch.no_grad():
        masks = gen.generate(image_np)

    if refine_masks:
        masks = _refine_sam_masks(masks)

    return sorted(masks, key=lambda x: x['area'], reverse=True)


def _refine_sam_masks(masks, open_k=3, close_k=5):
    refined = []
    for m in masks:
        seg = m['segmentation'].astype(np.uint8)
        k1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (open_k, open_k))
        k2 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (close_k, close_k))
        seg = cv2.morphologyEx(seg, cv2.MORPH_OPEN, k1)
        seg = cv2.morphologyEx(seg, cv2.MORPH_CLOSE, k2)
        new_area = int(seg.sum())
        if new_area > 0:
            m2 = dict(m)
            m2['segmentation'] = seg.astype(bool)
            m2['area'] = new_area
            refined.append(m2)
    return refined


def build_land_use_map(image_np, classified):
    h, w = image_np.shape[:2]
    lmap = np.zeros((h, w), dtype=np.uint8)

    # Layer by priority (lowest first, highest last)
    for seg in classified.get('open_land', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 0
    for seg in classified.get('road', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 0
    for seg in classified.get('vegetation', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 1
    for seg in classified.get('water', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 3
    for seg in classified.get('buildings', []):
        if seg['mask'].shape == (h, w):
            lmap[seg['mask']] = 2

    # Morphological cleanup
    k5 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
    bld_mask = (lmap == 2)
    veg_clean = cv2.morphologyEx((lmap == 1).astype(np.uint8), cv2.MORPH_OPEN, k5)
    veg_clean = cv2.morphologyEx(veg_clean, cv2.MORPH_CLOSE, k5)
    wat_clean = cv2.morphologyEx((lmap == 3).astype(np.uint8), cv2.MORPH_OPEN, k5)

    lmap[:] = 0
    lmap[veg_clean > 0] = 1
    lmap[wat_clean > 0] = 3
    lmap[bld_mask] = 2

    return lmap


print("SAM Auto + Land Use ready!")
print(f"   SAM config: points={SAM_CONFIG['points_per_side']}, "
      f"iou_thresh={SAM_CONFIG['pred_iou_thresh']}, "
      f"stability={SAM_CONFIG['stability_score_thresh']}")
```

```python
import cv2, numpy as np, uuid
from dataclasses import dataclass
from typing import List, Tuple, Dict
from PIL import Image as PILImage


@dataclass
class Building:
    building_id    : str
    bbox           : List[int]
    area_pixels    : int
    confidence     : float
    mask           : np.ndarray
    centroid       : Tuple[int, int]
    compactness    : float = 0.0
    rectangularity : float = 0.0
    solidity       : float = 0.0


def clean_masks(masks, opening=3, closing=7, min_area=120):
    out = []
    for m in masks:
        area = m.sum()
        u = m.astype(np.uint8) * 255
        op_sz = max(2, min(opening, int(np.sqrt(area) / 10)))
        cl_sz = max(3, min(closing, int(np.sqrt(area) / 5)))
        k1 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (op_sz, op_sz))
        k2 = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (cl_sz, cl_sz))
        u = cv2.morphologyEx(u, cv2.MORPH_OPEN, k1)
        u = cv2.morphologyEx(u, cv2.MORPH_CLOSE, k2)
        result = u > 0
        if result.sum() >= min_area:
            out.append(result)
    return out


def merge_overlapping(masks, iou_threshold=0.3):
    if len(masks) <= 1:
        return masks, 0
    merged = []
    used = set()
    n_merges = 0
    for i in range(len(masks)):
        if i in used:
            continue
        current = masks[i].copy()
        for j in range(i + 1, len(masks)):
            if j in used:
                continue
            inter = np.logical_and(current, masks[j]).sum()
            union = np.logical_or(current, masks[j]).sum()
            if inter / max(union, 1) > iou_threshold:
                current = np.logical_or(current, masks[j])
                used.add(j)
                n_merges += 1
        merged.append(current)
        used.add(i)
    return merged, n_merges


def filter_by_shape(segs, category='buildings'):
    kept = []
    removed = 0
    for seg in segs:
        shape = seg.get('shape', get_shape_info(seg['mask']))
        asp = shape.get('aspect_ratio', 1)
        sol = shape.get('solidity', 1)
        cmp = shape.get('compactness', 1)

        if category == 'buildings':
            # Reject elongated thin shapes (likely roads/paths)
            if asp > 10:
                removed += 1
                continue
            # Reject very low solidity (fragmented / non-compact)
            if sol < 0.30:
                removed += 1
                continue
            # Reject very low compactness (snake-like shapes)
            if cmp < 0.04:
                removed += 1
                continue

        kept.append(seg)
    return kept, removed


def register_buildings(classified_blds):
    blds = []
    for seg in classified_blds:
        mask = seg['mask']
        shape = seg.get('shape', get_shape_info(mask))
        if shape['area'] < 50:
            continue
        bid = f'BLD-{str(uuid.uuid4())[:6].upper()}'
        PILImage.fromarray(mask.astype(np.uint8) * 255).save(
            DIRS['masks'] / f'{bid}.png')
        blds.append(Building(
            building_id=bid, bbox=shape['bbox'],
            area_pixels=shape['area'],
            confidence=float(seg.get('clip_score', seg.get('quality', 0))),
            mask=mask, centroid=shape['centroid'],
            compactness=shape.get('compactness', 0),
            rectangularity=shape.get('rectangularity', 0),
            solidity=shape.get('solidity', 0),
        ))
    blds.sort(key=lambda b: b.area_pixels, reverse=True)
    return blds


print("Post-processing ready! (shape heuristics + IoU merge tracking)")
```

```python
from dataclasses import dataclass
from typing import List, Dict

@dataclass
class Neighborhood:
    grid_id: str
    row: int
    col: int
    bbox: List[int]
    n_buildings: int
    building_ids: List[str]
    total_pixels: int
    land_use_pct: Dict[str, float]
    density_class: str

def analyze_neighborhoods(land_map, buildings, grid_rows=3, grid_cols=3):
    h, w = land_map.shape
    cell_h, cell_w = h // grid_rows, w // grid_cols
    nbs = []
    for r in range(grid_rows):
        for c in range(grid_cols):
            y1 = r * cell_h
            y2 = (r + 1) * cell_h if r < grid_rows - 1 else h
            x1 = c * cell_w
            x2 = (c + 1) * cell_w if c < grid_cols - 1 else w
            cell = land_map[y1:y2, x1:x2]
            tp = cell.size
            pct = {
                'open': float(np.sum(cell == 0) / tp * 100),
                'vegetation': float(np.sum(cell == 1) / tp * 100),
                'built': float(np.sum(cell == 2) / tp * 100),
                'water': float(np.sum(cell == 3) / tp * 100),
            }
            bids = [b.building_id for b in buildings
                    if x1 <= b.centroid[0] < x2 and y1 <= b.centroid[1] < y2]
            density = 'high' if pct['built'] > 50 else 'medium' if pct['built'] > 20 else 'low'
            nbs.append(Neighborhood(
                grid_id=f"R{r}C{c}", row=r, col=c, bbox=[x1, y1, x2, y2],
                n_buildings=len(bids), building_ids=bids,
                total_pixels=tp, land_use_pct=pct, density_class=density))
    return nbs

def print_neighborhood_report(nbs):
    print("=" * 70)
    print("NEIGHBORHOOD ANALYSIS REPORT")
    print("=" * 70)
    for nb in nbs:
        p = nb.land_use_pct
        print(f"\n[{nb.grid_id}] {nb.density_class.upper()} | Bld: {nb.n_buildings} | "
              f"Built: {p['built']:.1f}% | Veg: {p['vegetation']:.1f}% | Open: {p['open']:.1f}%")
    print()

print("Neighborhood analysis ready!")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
plt.rcParams['font.family'] = 'DejaVu Sans'


def colored_overlay(image_np, masks, alpha=0.62):
    ov = image_np.copy().astype(float)
    cmap = plt.get_cmap('tab20')
    for i, m in enumerate(masks):
        if m.shape != image_np.shape[:2]: continue
        c = np.array(cmap(i % 20)[:3]) * 255
        for ch in range(3):
            ov[:, :, ch][m] = ov[:, :, ch][m] * (1 - alpha) + c[ch] * alpha
    return ov.astype(np.uint8)


def draw_hline(ax, y, xmin=0.05, xmax=0.95, color='#444', lw=1):
    ax.plot([xmin, xmax], [y, y], color=color, linewidth=lw,
            transform=ax.transAxes, clip_on=False, solid_capstyle='butt')


def visualize_results(image_np, classified, buildings, land_map,
                      neighborhoods=None, pipeline_stats=None,
                      title='UrbanInsight SAM+CLIP', save_path=None):
    fig = plt.figure(figsize=(26, 17))
    fig.patch.set_facecolor('#0d1117')
    n_b = len(buildings)
    n_v = len(classified.get('vegetation', []))
    n_w = len(classified.get('water', []))
    n_r = len(classified.get('road', []))
    n_u = len(classified.get('unknown', []))
    total_seg = sum(len(classified.get(c, [])) for c in
                    ['buildings', 'vegetation', 'road', 'water', 'open_land', 'unknown'])
    fig.suptitle(
        f'{title}  |  Buildings: {n_b}  Veg: {n_v}  Road: {n_r}  '
        f'Water: {n_w}  Unknown: {n_u}  Total: {total_seg}',
        fontsize=14, fontweight='bold', color='white', y=0.99)
    gs = fig.add_gridspec(2, 3, hspace=0.28, wspace=0.14,
                          left=0.02, right=0.98, top=0.94, bottom=0.02)
    axes = [fig.add_subplot(gs[r, c]) for r in range(2) for c in range(3)]
    h, w = image_np.shape[:2]
    axes[0].imshow(image_np)
    axes[0].set_title('Original Image', color='white', fontsize=13, pad=8)

    CAT_COLORS = {'buildings': (220,70,50), 'vegetation': (50,195,70),
                  'road': (150,150,150), 'water': (60,140,220),
                  'open_land': (180,155,90), 'unknown': (100,100,100)}
    ov1 = image_np.copy().astype(float)
    leg1 = []
    for cat, (cr, cg, cb) in CAT_COLORS.items():
        segs = classified.get(cat, [])
        if not segs: continue
        c = np.array([cr, cg, cb], dtype=float)
        for seg in segs:
            m = seg['mask']
            if m.shape != (h, w): continue
            for ch in range(3):
                ov1[:, :, ch][m] = ov1[:, :, ch][m] * 0.25 + c[ch] * 0.75
        leg1.append(mpatches.Patch(color=np.array([cr,cg,cb])/255,
                                   label=f'{cat} ({len(segs)})'))
    axes[1].imshow(ov1.astype(np.uint8))
    if leg1:
        axes[1].legend(handles=leg1, loc='upper right', fontsize=8,
                       framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[1].set_title(f'SAM+CLIP Classification -- {total_seg} segments',
                      color='white', fontsize=13, pad=8)

    bld_masks = [b.mask for b in buildings]
    axes[2].imshow(colored_overlay(image_np, bld_masks, 0.70))
    axes[2].set_title(f'Building Masks -- {n_b}', color='white', fontsize=13, pad=8)

    lrgb = np.zeros((h, w, 3), dtype=np.uint8)
    lrgb[land_map == 0] = [90, 110, 200]
    lrgb[land_map == 1] = [50, 190, 70]
    lrgb[land_map == 2] = [210, 55, 55]
    lrgb[land_map == 3] = [55, 145, 220]
    total_px = land_map.size
    pct = {k: np.sum(land_map == v) / total_px * 100
           for k, v in {'open': 0, 'veg': 1, 'built': 2, 'water': 3}.items()}
    axes[3].imshow(lrgb)
    lu_leg = [
        mpatches.Patch(color='#5A6EC8', label=f"Open/Road {pct['open']:.1f}%"),
        mpatches.Patch(color='#32BE46', label=f"Vegetation {pct['veg']:.1f}%"),
        mpatches.Patch(color='#D23737', label=f"Built-up {pct['built']:.1f}%"),
        mpatches.Patch(color='#3791DC', label=f"Water {pct['water']:.1f}%"),
    ]
    axes[3].legend(handles=lu_leg, loc='upper right', fontsize=9,
                   framealpha=0.85, facecolor='#111', labelcolor='white')
    axes[3].set_title('Land Use Map (SAM+CLIP)', color='white', fontsize=13, pad=8)

    axes[4].imshow(colored_overlay(image_np, bld_masks, 0.55))
    for b in buildings:
        cx, cy = b.centroid
        fs = max(5, min(9, int(b.area_pixels**0.38 * 0.30)))
        axes[4].text(cx, cy, b.building_id[-6:], color='white', fontsize=fs,
                     ha='center', va='center', fontweight='bold',
                     bbox=dict(boxstyle='round,pad=0.15', fc='black', alpha=0.65, ec='none'))
        x1, y1, x2, y2 = b.bbox
        axes[4].add_patch(mpatches.FancyBboxPatch(
            (x1, y1), (x2-x1), (y2-y1), boxstyle='square,pad=0',
            lw=1.5, edgecolor='yellow', facecolor='none'))
    if neighborhoods:
        for nb in neighborhoods:
            x1, y1, x2, y2 = nb.bbox
            clr = {'high': '#ff4444', 'medium': '#ffaa00', 'low': '#44ff44'}[nb.density_class]
            axes[4].add_patch(mpatches.Rectangle(
                (x1, y1), x2-x1, y2-y1, linewidth=2, edgecolor=clr,
                facecolor='none', linestyle='--', alpha=0.7))
            axes[4].text(x1+5, y1+15, nb.grid_id, color=clr, fontsize=8,
                         fontweight='bold',
                         bbox=dict(fc='black', alpha=0.5, ec='none', pad=1))
    axes[4].set_title(f'Building IDs -- {n_b} registered', color='white', fontsize=13, pad=8)

    ax5 = axes[5]
    ax5.set_facecolor('#161b22')
    ax5.axis('off')
    ax5.set_xlim(0, 1); ax5.set_ylim(0, 1)

    def T(x, y, s, sz=11, col='white', fw='normal', ha='center'):
        ax5.text(x, y, s, transform=ax5.transAxes, fontsize=sz, color=col,
                 ha=ha, va='center', fontweight=fw)

    avg_a = int(np.mean([b.area_pixels for b in buildings])) if buildings else 0
    avg_c = float(np.mean([b.confidence for b in buildings])) if buildings else 0

    # Pipeline stats
    stats = pipeline_stats or {}
    removed_area  = stats.get('removed_by_area', '--')
    shape_removed = stats.get('shape_filtered', '--')
    n_merges      = stats.get('n_merges', '--')

    T(0.5, 0.97, 'Scene Statistics', 15, 'white', 'bold')
    draw_hline(ax5, 0.91)

    info = [
        ('Buildings detected',   f'{n_b}',         '#FFB347'),
        ('Vegetation segments',  f'{n_v}',         '#32BE46'),
        ('Road segments',        f'{n_r}',         '#888888'),
        ('Water segments',       f'{n_w}',         '#3791DC'),
        ('Unknown segments',     f'{n_u}',         '#AAAAAA'),
        ('Image resolution',     f'{w} x {h} px',  '#87CEEB'),
        ('Avg building area',    f'{avg_a:,} px',   '#87CEEB'),
        ('Avg CLIP confidence',  f'{avg_c:.3f}',   '#FFB347'),
    ]
    for i2, (k, v, col) in enumerate(info):
        yp = 0.85 - i2 * 0.065
        T(0.05, yp, k, 9, '#AAA', ha='left')
        T(0.95, yp, v, 9, col, ha='right')

    draw_hline(ax5, 0.31)
    T(0.5, 0.27, 'Pipeline Stats', 12, 'white', 'bold')
    pipeline_info = [
        ('Removed by area filter',  str(removed_area),  '#FFAAAA'),
        ('Removed by shape filter', str(shape_removed), '#FFAAAA'),
        ('Merges performed',        str(n_merges),      '#AAFFAA'),
    ]
    for i3, (k, v, col) in enumerate(pipeline_info):
        yp = 0.21 - i3 * 0.055
        T(0.05, yp, k, 9, '#AAA', ha='left')
        T(0.95, yp, v, 9, col, ha='right')

    draw_hline(ax5, 0.05)
    T(0.5, 0.025, 'UrbanInsight  |  SAM Auto + CLIP Classifier  (Improved)', 8, '#555')

    for ax in axes[:5]:
        ax.set_facecolor('#0d1117')
        for sp in ax.spines.values():
            sp.set_edgecolor('#30363d'); sp.set_linewidth(1.2)
        ax.tick_params(colors='#555', labelsize=7)

    if save_path:
        plt.savefig(save_path, dpi=140, bbox_inches='tight',
                    facecolor='#0d1117', edgecolor='none')
        print(f'Saved: {save_path}')
    plt.show()


print("Visualization ready!")
```

```python
# ================================================================
# FULL PIPELINE: SAM Auto + CLIP Classification  (محسّن)
# ================================================================

import time, json, numpy as np
from PIL import Image
from pathlib import Path


def run_pipeline(image_input,
                 points_per_side=None,
                 min_bld_area=200,
                 neighborhood_grid=(3, 3),
                 save_name='result'):
    t0 = time.time()

    if isinstance(image_input, (str, Path)):
        img = np.array(Image.open(str(image_input)).convert('RGB'))
    else:
        img = np.array(image_input.convert('RGB'))

    h, w = img.shape[:2]
    print(f"Image: {w} x {h} px")

    MAX_DIM = 1200
    scale = 1.0
    if max(h, w) > MAX_DIM:
        scale = MAX_DIM / max(h, w)
        nh, nw = int(h * scale), int(w * scale)
        img = np.array(Image.fromarray(img).resize((nw, nh), Image.LANCZOS))
        h, w = nh, nw
        print(f"Resized: {nw} x {nh}")

    adj_min = max(50, int(min_bld_area * scale**2))
    pts = points_per_side if points_per_side is not None else SAM_CONFIG['points_per_side']

    # Step 1: SAM Auto
    print(f"\n[1/6] SAM Auto segmentation (points={pts})...")
    sam_masks = run_sam_auto(img,
                             points_per_side=pts,
                             min_area=max(50, adj_min // 4),
                             refine_masks=True)
    print(f"  Raw segments: {len(sam_masks)}")

    # Step 2: CLIP classification
    print(f"\n[2/6] CLIP classification (ensemble + confidence threshold)...")
    classified = classify_segments_batch(img, sam_masks, batch_size=16)
    removed_by_area = classified.pop('_stats', {}).get('removed_by_area', 0)

    # Step 3: Post-process buildings
    print(f"\n[3/6] Post-processing buildings...")
    raw_bld_segs = classified['buildings']
    bld_segs_ok, shape_removed = filter_by_shape(raw_bld_segs, category='buildings')
    print(f"  Shape filter: {len(raw_bld_segs)} -> {len(bld_segs_ok)} "
          f"(-{shape_removed} by shape)")
    raw_masks = [s['mask'] for s in bld_segs_ok]
    cl_masks = clean_masks(raw_masks, opening=3, closing=7,
                            min_area=max(50, adj_min // 3))
    cl_masks, n_merges = merge_overlapping(cl_masks, iou_threshold=0.3)
    print(f"  {len(raw_masks)} -> {len(cl_masks)} after cleaning "
          f"({n_merges} merges)")

    classified['buildings'] = []
    for m in cl_masks:
        sh = get_shape_info(m)
        classified['buildings'].append({
            'mask': m, 'area': int(m.sum()), 'quality': 1.0,
            'clip_score': 1.0, 'bbox': sh['bbox'],
            'centroid': sh['centroid'], 'shape': sh,
        })

    # Step 4: Land use map
    print(f"\n[4/6] Building land use map...")
    land_map = build_land_use_map(img, classified)
    total = land_map.size
    open_pct = np.sum(land_map==0)/total*100
    veg_pct  = np.sum(land_map==1)/total*100
    blt_pct  = np.sum(land_map==2)/total*100
    wat_pct  = np.sum(land_map==3)/total*100
    print(f"  Open: {open_pct:.1f}%  Veg: {veg_pct:.1f}%  "
          f"Built: {blt_pct:.1f}%  Water: {wat_pct:.1f}%")

    # Step 5: Register
    print(f"\n[5/6] Registering buildings...")
    buildings = register_buildings(classified['buildings'])
    print(f"  Registered: {len(buildings)} buildings")

    # Step 6: Neighborhoods
    print(f"\n[6/6] Neighborhood analysis...")
    grid_r, grid_c = neighborhood_grid
    neighborhoods = analyze_neighborhoods(land_map, buildings,
                                          grid_rows=grid_r, grid_cols=grid_c)
    print_neighborhood_report(neighborhoods)

    elapsed = time.time() - t0
    print(f"\nTotal time: {elapsed:.1f}s")

    pipeline_stats = {
        'removed_by_area': removed_by_area,
        'shape_filtered': shape_removed,
        'n_merges': n_merges,
        'n_buildings': len(buildings),
        'n_vegetation': len(classified.get('vegetation', [])),
        'n_road': len(classified.get('road', [])),
        'n_water': len(classified.get('water', [])),
        'n_unknown': len(classified.get('unknown', [])),
    }

    save_path = str(DIRS['vis'] / f'{save_name}.png')
    visualize_results(img, classified, buildings, land_map,
                      neighborhoods=neighborhoods,
                      pipeline_stats=pipeline_stats,
                      save_path=save_path)

    # Save JSONs
    reg = [{'id': b.building_id, 'bbox': b.bbox, 'area_px': b.area_pixels,
             'confidence': round(b.confidence, 4), 'centroid': b.centroid,
             'compactness': round(b.compactness, 4),
             'rectangularity': round(b.rectangularity, 4),
             'solidity': round(b.solidity, 4),
             'mask_file': f'{b.building_id}.png'} for b in buildings]
    with open(DIRS['results'] / f'{save_name}_buildings.json', 'w') as f:
        json.dump(reg, f, indent=2)

    nb_data = [{'grid_id': nb.grid_id, 'bbox': nb.bbox,
                'n_buildings': nb.n_buildings, 'building_ids': nb.building_ids,
                'land_use_pct': nb.land_use_pct,
                'density_class': nb.density_class} for nb in neighborhoods]
    with open(DIRS['results'] / f'{save_name}_neighborhoods.json', 'w') as f:
        json.dump(nb_data, f, indent=2)

    stats_dict = {
        'pipeline': 'SAM Auto + CLIP (Improved)',
        'sam_config': SAM_CONFIG,
        'clip_config': CLIP_CONFIG,
        'n_buildings': len(buildings),
        'n_vegetation': len(classified.get('vegetation', [])),
        'n_road': len(classified.get('road', [])),
        'n_water': len(classified.get('water', [])),
        'n_unknown': len(classified.get('unknown', [])),
        'pipeline_stats': pipeline_stats,
        'land_use': {
            'open_road_pct': round(open_pct, 2),
            'veg_pct':       round(veg_pct,  2),
            'built_pct':     round(blt_pct,  2),
            'water_pct':     round(wat_pct,  2),
        },
        'neighborhoods': nb_data,
        'elapsed_s': round(elapsed, 2),
    }
    with open(DIRS['results'] / f'{save_name}_stats.json', 'w') as f:
        json.dump(stats_dict, f, indent=2)

    print(f"\nFiles saved to {DIRS['results']}")
    return {'image': img, 'buildings': buildings, 'classified': classified,
            'land_use_map': land_map, 'neighborhoods': neighborhoods,
            'pipeline_stats': pipeline_stats}


print("Pipeline ready: SAM Auto + CLIP (Improved)")
print("Usage: results = run_pipeline(image_path_or_pil_image)")
```
